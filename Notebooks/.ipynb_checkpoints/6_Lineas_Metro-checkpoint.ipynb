# Carga de datos

```python
#lectura shapefile Paradas Metro
datos = gpd.read_file('C:/Users/pc_vi/OneDrive/Escritorio/TFG/Notebooks/shapefiles/MetroMerged.shp')
#sistema de coordenadas y proyección
datos.crs = {'init':'epsg:25830'}  #ETRS89 UTM Zona 30N (Madrid)
#se borran columnas innecesarias
datos = datos.drop(['FECHAACTUA','MODO','SITUACION','CODIGOCTME','DENOMINA_1','MODOINTERC','CODIGOINTE','TIPO','CODIGOPROV','CODIGOMUNI',
                   'CODIGOENTI','CODIGONUCL','CODIGOVIA','TIPOVIA','PARTICULA','NOMBREVIA','TIPONUMERO','NUMEROPORT','CALIFICADO','CARRETERA',
                   'CODIGOPOST','SECCIONCEN','TESELA','SECTORURBA','SECTOR','CORREDOR','CORONATARI','CORONA123','ZONATRANSP','ENCUESTADO',
                   'ENCUESTAAF','HOJA25000','ACONDICION','ACONDICI_1','FECHAALTA','FECHAINICI','FECHAFIN','GRADOACCES','SITUACIONC','DENOMINA_2',
                   'INTERURBAN','INTERURB_1'], axis=1)
datos.explore('DENOMINACI',legend=False)
```

```python
print("Filas, Columnas:",datos.shape)
print(datos.duplicated(keep=False).sum(),"datos duplicados")
```

```python
#Se renombran algunas columnas:
datos.rename(columns={'Utilizacio': 'Utilizaciones2014','Utilizac_1': 'Utilizaciones2015','Utilizac_2': 'Utilizaciones2016',
                     'Utilizac_3': 'Utilizaciones2017','Utilizac_4': 'Utilizaciones2018','Utilizac_5': 'Utilizaciones2019',
                     'Utilizac_6': 'Utilizaciones2020','Utilizac_7': 'Utilizaciones2021','Utilizac_8': 'Utilizaciones2022',
                     'Utilizac_9': 'Utilizaciones2023'}, inplace=True)
```

```python
lineas = [linea1,linea2,linea3,linea4,linea5,linea6,linea7,linea8,linea9,linea10,linea11,linea12,lineaR]

utilizacion_total = {}

for i, linea in enumerate(lineas):
    suma_utilizaciones = linea.loc[:, 'Utilizaciones2014':'Utilizaciones2023'].sum().sum()
    nombre_linea = f"Línea {i + 1}"
    utilizacion_total[nombre_linea] = suma_utilizaciones

lineas_ordenadas = sorted(utilizacion_total.items(), key=lambda x: x[1], reverse=True)

nombres_lineas_ordenadas = [x[0] for x in lineas_ordenadas]
sumas_ordenadas = [x[1] for x in lineas_ordenadas]

plt.figure(figsize=(10, 6))
plt.bar(range(len(nombres_lineas_ordenadas)), sumas_ordenadas, align='center')
plt.xticks(range(len(nombres_lineas_ordenadas)), nombres_lineas_ordenadas, rotation=45, ha='right')
plt.xlabel('Línea de metro')
plt.ylabel('Utilización total')
plt.title('Utilización total por línea de metro')
plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```

```python
for year in range(2014, 2024):
    utilizacion_total = {}

    for i, linea in enumerate(lineas):
        suma_utilizaciones = linea[f'Utilizaciones{year}'].sum()
        nombre_linea = f"Línea {i + 1}"
        utilizacion_total[nombre_linea] = suma_utilizaciones

    lineas_ordenadas = sorted(utilizacion_total.items(), key=lambda x: x[1], reverse=True)

    nombres_lineas_ordenadas = [x[0] for x in lineas_ordenadas]
    sumas_ordenadas = [x[1] for x in lineas_ordenadas]

    plt.figure(figsize=(10, 6))
    plt.bar(range(len(nombres_lineas_ordenadas)), sumas_ordenadas, align='center')
    plt.xticks(range(len(nombres_lineas_ordenadas)), nombres_lineas_ordenadas, rotation=45, ha='right')
    plt.xlabel('Línea de metro')
    plt.ylabel('Utilización total')
    plt.title(f'Utilización total por línea de metro en el año {year}')
    plt.ticklabel_format(style='plain', axis='y')
    plt.tight_layout()
    plt.show()
```
